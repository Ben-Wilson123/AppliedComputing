### FASTA Comparison

```python
with open("../data/cleaned_sequences_AI.fasta") as f:
    ai_headers = [line.strip() for line in f if line.startswith(">")]

with open("../data/cleaned_sequences_regex.fasta") as f:
    regex_headers = [line.strip() for line in f if line.startswith(">")]

print(f"{'AI':<12} {'Regex':<12} {'Agree?'}")
print("-" * 36)

for ai, regex in zip(ai_headers, regex_headers):
    print(f"{ai:<12} {regex:<12} {ai == regex}")
```

##### The AI and Regex files only on 3/8 cases. Part of this write-up was to understand the different ways in which they fail. The Regex file has trouble parsing through the species name. The messy file makes it difficult to amend the abbreviations for the homo genus, which is why many of the headers still say H sapiens or H. sapiens. Another way in which the regex file fails is through incosistent naming conventions for the sample number. Some of the samples do not possess the double zeroes before their number, so one sample may say "sample_007" while another says "sample_8". These errors lie in the defined patterns for species and samples. I was not able to determine errors with the AI extraction - sample names are standardized and consisten, the organisms are all the proper formatting for Homo Sapiens, and all of the genes are capitalized and correctly outputted.
